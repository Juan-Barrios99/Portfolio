```python
#STEP 1: Import libraries
import pandas as pd
import numpy as np
```

```python
#STEP 2: Load the dataset
df = pd.read_csv("../data/datasales.csv", encoding="cp1252")
```

```python
#STEP 3: Check for missing values and info
df.info()
```

```python
#STEP 3.5: Handle missing values (NaN)
df.isnull().sum()
```

```python
#Ensure numeric colums arenumeric before filling NaN
df["Row ID"] = pd.to_numeric(
    df["Row ID"],
    errors="coerce"
)

df["Postal Code"] = pd.to_numeric(
    df["Postal Code"],
    errors="coerce"
)

df["Sales"] = pd.to_numeric(
    df["Sales"],
    errors="coerce"
)

df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["Profit"] = pd.to_numeric(
    df["Profit"],
    errors="coerce"
)
```

```python
#Fill missing numeric values with the column mean
df["Row ID"] = df["Row ID"].fillna(df["Row ID"].mean())
df["Postal Code"] = df["Postal Code"].fillna(df["Postal Code"].mean())
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Profit"] = df["Profit"].fillna(df["Profit"].mean())
```

```python
#Make sure there are no more missing values
df.isnull().sum()
```

```python
#STEP 4: Clean column names (remove leading/trailing spaces)
df.columns = df.columns.str.strip()
```

```python
#STEP 5: Fix date formats(turn inconsistent text into real date format)
df["Order Date"]=pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")
```

```python
#STEP 6: Standarize text columns(make names consistent and neat)

for col in ["Order ID", "Ship Mode", "Customer ID",
            "Customer Name", "Segment", "Country", "City",
            "State", "Region", "Product ID", "Category",
            ]:
    
    df[col] = df[col].str.strip().str.title()
```

```python
#STEP 7: Clean currency columns (Unit price, total sales)
for col in ["Sales", "Profit"]:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(r"[$,USD ]", "", regex=True)
        .str.strip()
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
#STEP 8:Fix Discount(%)-Some values are like 10%, 
# 0.10 or 8 percent- letsmake them consistent
df["Discount"] = (
    df["Discount"]
    .astype(str)
    .str.replace("%", "", regex=True)
    .str.replace("percent", "", regex=True)
    .astype(float)
)

df["Discount"] = np.where(
    df["Discount"] < 1,
    df["Discount"] * 100,
    df["Discount"]
)
```

```python
df.isnull().sum()
```

```python
#STEP 9

# Delete duplicate rows
df.drop_duplicates(inplace=True)

# Delete rows with missing key data (clean final dataset)
df.dropna(inplace=True)

#Reset the index after dropping rows
df.reset_index(drop=True, inplace=True)
```
